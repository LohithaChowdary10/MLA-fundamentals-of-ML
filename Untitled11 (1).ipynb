```python
import csv

a = []

with open('enjoysport.csv', 'r') as csvfile:
    for row in csv.reader(csvfile):
        a.append(row)

print(a)

print("\nThe total number of training instances are:", len(a))

num_attribute = len(a[0]) - 1

print("\nThe initial hypothesis is:")

hypothesis = ['0'] * num_attribute

print(hypothesis)

for i in range(0, len(a)):
    if a[i][num_attribute].lower() == 'yes':
        for j in range(0, num_attribute):
            if hypothesis[j] == '0':
                hypothesis[j] = a[i][j]
            elif hypothesis[j] != a[i][j]:
                hypothesis[j] = '?'

    print("\nThe hypothesis for the training instance {} is:\n{}".format(i + 1, hypothesis))

print("\nThe Maximally Specific Hypothesis is:")
print(hypothesis)
```

```python
import numpy as np
import pandas as pd

# Read the dataset
data = pd.read_csv('enjoysport.csv')

# Separate concepts and target
concepts = np.array(data.iloc[:, 0:-1])
target = np.array(data.iloc[:, -1])

print("Training Data:")
print(concepts)

print("\nTarget:")
print(target)


def learn(concepts, target):

    # Initialize specific hypothesis
    specific_h = concepts[0].copy()

    print("\nInitial Specific Hypothesis:")
    print(specific_h)

    # Initialize general hypothesis
    general_h = [["?" for i in range(len(specific_h))]
                 for i in range(len(specific_h))]

    print("\nInitial General Hypothesis:")
    print(general_h)

    # Candidate Elimination
    for i, h in enumerate(concepts):

        # For positive example
        if target[i] == "yes":

            for x in range(len(specific_h)):

                if h[x] != specific_h[x]:
                    specific_h[x] = '?'
                    general_h[x][x] = '?'

        # For negative example
        if target[i] == "no":

            for x in range(len(specific_h)):

                if h[x] != specific_h[x]:
                    general_h[x][x] = specific_h[x]
                else:
                    general_h[x][x] = '?'

        print("\nStep", i + 1)
        print("Specific Hypothesis:")
        print(specific_h)

        print("General Hypothesis:")
        print(general_h)

    # Remove completely general hypotheses
    indices = [
        i for i, val in enumerate(general_h)
        if val == ['?', '?', '?', '?', '?', '?']
    ]

    for i in reversed(indices):
        general_h.pop(i)

    return specific_h, general_h


# Run Candidate Elimination
s_final, g_final = learn(concepts, target)


# Final output
print("\n====================================")
print("FINAL RESULT")
print("====================================")

print("\nFinal Specific Hypothesis:")
print(s_final)

print("\nFinal General Hypothesis:")
print(g_final)
```
